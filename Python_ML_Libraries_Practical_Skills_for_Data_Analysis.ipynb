```python
# No1 NUMPY reshape numpy 1d array into 2d ,3d
import numpy as np

# Create 1D array from 1 to 21
arr_1d = np.arange(1, 21)
print("1D Array:")
print(arr_1d)

# Reshape to 2D (5,4)
arr_2d = arr_1d.reshape(5, 4)
print("\n2D Array (5x4):")
print(arr_2d)

# Reshape to 3D (2,5,2)
arr_3d = arr_1d.reshape(2, 5, 2)
print("\n3D Array (2x5x3):")
print(arr_3d)
```

```python
#no 2 Pandas

import pandas as pd

# Create two Series
s1 = pd.Series([10, 20, 30], index=['a', 'b', 'c'])
s2 = pd.Series([2, 4, 5], index=['a', 'b', 'c'])

# Division
result = s1 / s2
print("Division Result:")
print(result)
```

```python
#no3 matplotlib
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

plt.plot(x, y, color='green')
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("Green  Line Graph")
plt.show()
```

```python
#no 4 scipy
import numpy as np
from scipy import linalg

# Define matrix
A = np.array([[2, 1],
              [1, 2]])

# Determinant
det = linalg.det(A)
print("Determinant:", det)

# Eigenvalues and Eigenvectors
eigenvalues, eigenvectors = linalg.eig(A)
print("Eigenvalues:", eigenvalues)
print("Eigenvectors:\n", eigenvectors)
```

```python
import numpy as np
import statistics

data = [10, 20, 20, 30, 40]

mean = np.mean(data)
median = np.median(data)
mode = statistics.mode(data)
variance = np.var(data)
std_dev = np.std(data)

print("Mean:", mean)
print("Median:", median)
print("Mode:", mode)
print("Variance:", variance)
print("Standard Deviation:", std_dev)
```
